## 2.2 Numerical Feature Analysis

### Budget and Revenue Relationship

通过散点图观察电影预算（Budget）与票房收入（Revenue）的关系。

结果显示，预算较高的电影通常具有更高的票房收入，但仍存在部分高预算低收益电影，说明预算并不是决定票房成功的唯一因素。

### Popularity and Revenue Relationship

分析电影受欢迎程度（Popularity）与票房收入之间的关系。

结果发现，Popularity 与 Revenue 存在一定正相关关系，但并非所有高热度电影都会获得高票房。

### Vote Average and Revenue Relationship

分析评分（Vote Average）与票房收入之间的关系。

结果表明，高评分电影并不一定拥有最高票房收入，商业成功还受到市场推广、演员阵容、制作规模等因素影响。

```python
movies["genres"].head()
```

```python
import ast

movies["genres"] = movies["genres"].apply(ast.literal_eval)
```

```python
movies["genres"].iloc[0]
```

```python
genres = []

for movie_genres in movies["genres"]:
    for genre in movie_genres:
        genres.append(genre["name"])

genres[:10]
```

```python
from collections import Counter

genre_counts = Counter(genres)

genre_counts
```

```python
genre_df = pd.DataFrame(
    genre_counts.items(),
    columns=["Genre", "Count"]
)

genre_df.head()
```

```python
genre_df = genre_df.sort_values(
    by="Count",
    ascending=False
)

genre_df.head(10)
```

```python
genre_df = genre_df.reset_index(drop=True)

genre_df.head(10)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.bar(
    genre_df["Genre"].head(10),
    genre_df["Count"].head(10)
)

plt.xticks(rotation=45)

plt.title("Top 10 Movie Genres")

plt.xlabel("Genre")
plt.ylabel("Number of Movies")

plt.show()
```



## 2.3 Movie Genre Analysis

对电影类型字段进行统计分析，并观察不同类型电影在数据集中的分布情况。

结果显示，Drama 类型电影数量最多，其次是 Comedy 和 Thriller。
Action、Romance 等类型也占有较高比例。

该结果说明，该数据集中不同电影类型的数量分布存在明显差异，Drama 和 Comedy 类型电影占据较大比例。

## Summary

综合以上探索性分析可以发现：

电影票房收入（Revenue）与预算（Budget）、受欢迎程度（Popularity）以及评分（Vote Average）之间存在一定关系，但没有单一因素能够完全决定电影的商业成功。

高预算、高热度电影通常具有更高票房潜力，但部分电影仍可能出现投入较高但收益较低的情况。

此外，电影类型分布显示 Drama、Comedy 和 Thriller 等类型在数据集中占比较高。

总体来看，电影商业表现受到多种因素共同影响，包括制作规模、市场推广、观众兴趣以及电影类型等。
