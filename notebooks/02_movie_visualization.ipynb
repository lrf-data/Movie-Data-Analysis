- Explore genre distribution.
- Identify movies with high revenue.
- Compare popularity, ratings, and commercial performance.

```python
import pandas as pd

movies = pd.read_csv("../data/tmdb_5000_movies.csv")

movies.head()
```

```python
movies["release_date"].head()
```

```python
movies["release_date"] = pd.to_datetime(movies["release_date"])

movies["release_year"] = movies["release_date"].dt.year

movies[["release_date", "release_year"]].head()
```

```python
movies_per_year = movies["release_year"].value_counts().sort_index()

movies_per_year.head()
```

```python
.sort_index()
```

```python
movies_per_year
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(12, 6))

plt.plot(movies_per_year.index, movies_per_year.values)

plt.xlabel("Release Year")
plt.ylabel("Number of Movies")
plt.title("Number of Movies Released Over Years")

plt.show()
```

## Analysis: Movie Release Trend Over Years

The visualization shows the number of movies released each year.

From the chart, we can observe that movie production has generally increased over time.

The number of movies released was relatively low in earlier years, while it increased significantly after the 2000s.

This suggests that the film industry expanded rapidly in recent decades.

However, the increase in movie quantity does not necessarily mean every movie achieved commercial success.

```python
movies["genres"].head()
```

```python
import ast

movies["genres"] = movies["genres"].apply(ast.literal_eval)

movies["genres"].head()
```

```python
def extract_genres(genre_list):
    return [genre["name"] for genre in genre_list]

movies["genre_names"] = movies["genres"].apply(extract_genres)

movies["genre_names"].head()
```

```python
genre_data = movies.explode("genre_names")

genre_data["genre_names"].head()
```

```python
genre_counts = genre_data["genre_names"].value_counts()

genre_counts.head()
```

```python
plt.figure(figsize=(10, 6))

genre_counts.head(10).plot(kind="bar")

plt.xlabel("Genre")
plt.ylabel("Number of Movies")
plt.title("Top 10 Movie Genres")

plt.xticks(rotation=45)

plt.show()
```

## Analysis: Movie Genre Distribution

The visualization shows the distribution of movie genres in the dataset.

Drama is the most common genre, followed by Comedy and Thriller.

This indicates that drama and comedy movies represent a large proportion of the movie industry in this dataset.

However, one movie can belong to multiple genres, so the total genre count is larger than the number of movies.

```python
movies[["title", "revenue"]].head()
```

```python
top_revenue_movies = movies.sort_values(
    by="revenue",
    ascending=False
)

top_revenue_movies[["title", "revenue"]].head(10)
```

```python
plt.figure(figsize=(10, 6))

top10_revenue = top_revenue_movies.head(10)

plt.barh(
    top10_revenue["title"],
    top10_revenue["revenue"]
)

plt.xlabel("Revenue")
plt.ylabel("Movie Title")
plt.title("Top 10 Movies by Revenue")

plt.gca().invert_yaxis()

plt.show()
```

## Analysis: Top Revenue Movies

The visualization shows the top 10 movies with the highest revenue in the dataset.

Most of these movies are large-scale commercial productions with global audiences.

Many top revenue movies belong to popular genres such as Action, Adventure, and Science Fiction.

This suggests that commercial success is influenced by multiple factors, including production scale, audience reach, and marketing strategies.

```python
movies[["title", "popularity", "revenue"]].head()
```

```python
top_popularity_movies = movies.sort_values(
    by="popularity",
    ascending=False
)

top_popularity_movies[["title", "popularity", "revenue"]].head(10)
```

```python
plt.figure(figsize=(10, 6))

top10_popularity = top_popularity_movies.head(10)

plt.barh(
    top10_popularity["title"],
    top10_popularity["popularity"]
)

plt.xlabel("Popularity")
plt.ylabel("Movie Title")
plt.title("Top 10 Movies by Popularity")

plt.gca().invert_yaxis()

plt.show()
```

## Analysis: Popularity vs Revenue

The comparison shows that movies with high popularity do not always generate the highest revenue.

Some movies attract significant public attention but do not achieve the highest commercial returns.

This suggests that popularity is only one factor affecting box office performance. Other factors, such as production budget, marketing strategy, release scale, and audience reach, also influence commercial success.

```python
movies[["title", "vote_average", "revenue"]].head()
```

```python
top_rated_movies = movies.sort_values(
    by="vote_average",
    ascending=False
)

top_rated_movies[["title", "vote_average", "revenue"]].head(10)
```

```python
rated_movies = movies[movies["revenue"] > 0]

rated_movies.shape
```

```python
top_rated_revenue_movies = rated_movies.sort_values(
    by="vote_average",
    ascending=False
)

top_rated_revenue_movies[["title", "vote_average", "revenue"]].head(10)
```

```python
plt.figure(figsize=(10, 6))

top10_rated = top_rated_revenue_movies.head(10)

plt.barh(
    top10_rated["title"],
    top10_rated["revenue"]
)

plt.xlabel("Revenue")
plt.ylabel("Movie Title")
plt.title("Revenue of Top Rated Movies")

plt.gca().invert_yaxis()

plt.show()
```

## Analysis: Rating vs Revenue

The visualization shows the revenue performance of highly rated movies.

Although some highly rated movies achieved strong commercial success, there are also highly rated movies with relatively low revenue.

This suggests that movie ratings and commercial success are related but not determined by each other.

Other factors, such as production budget, marketing strategy, audience size, and distribution scale, also influence revenue.
